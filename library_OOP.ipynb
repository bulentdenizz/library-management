```python
class Book:

    def __init__(self, ad, yazar, sayfa_sayisi, yil):
        self._ad = ad
        self._yazar = yazar
        self._sayfa_sayisi = sayfa_sayisi
        self._yil = yil
        self._durum = "Müsait"

    def get_ad(self):
        return self._ad

    def set_durum(self, yeni_durum):
        self._durum = yeni_durum
        print(f"'{self._ad}' durumu: {yeni_durum}.")

    def __str__(self):
        return f"Kitap Adı: {self._ad:<20} | Yazar: {self._yazar:<15} | Durum: {self._durum}"

    def display_info(self):
        return f"({self._yil}) {self._ad} - Sayfa: {self._sayfa_sayisi}"
```

```python
class Person:

    def __init__(self, ad, soyad):
        self._ad = ad
        self._soyad = soyad
        self._kitaplar = []

    def display_info(self):
        return f"Kişi: {self._ad} {self._soyad}"

    def odunc_al(self,kitap_adi):
        self._kitaplar.append(kitap_adi)

class User(Person):

    def __init__(self, ad, soyad, ogrenci_no):
        super().__init__(ad, soyad)
        self._ogrenci_no = ogrenci_no

    def display_info(self):
        return f"Öğrenci: {self._ad} {self._soyad} (No: {self._ogrenci_no})"

class Admin(Person):

    def __init__(self, ad, soyad, yetki_seviyesi):
        super().__init__(ad, soyad)
        self._yetki = yetki_seviyesi
        self._gorev = "Kütüphane Görevlisi"

    def display_info(self):
        return f"Admin: {self._ad} {self._soyad} | Yetki: {self._yetki}"

    def kitap_ekle_yetkisi(self):
        print("Admin yetkisi ile yeni kitap eklenebilir.")
```

```python
class Library:
    def __init__(self):
        self.envanter = []
        self.kullanicilar = {}
        self.toplam_kitap = 0

    def kitap_ekle(self, book_object):
        if isinstance (book_object, Book):
            self.envanter.append(book_object)
            self.toplam_kitap += 1
            print(f"'{book_object.get_ad()}' kütüphaneye eklendi.")
        else:
            print("HATA: Sadece Kitap tipinde nesneler eklenebilir.")

    def kitap_sil(self, kitap_adi):
        silinecek_kitap = None

        for kitap in self.envanter:

            if kitap.get_ad().lower() == kitap_adi.lower():
                silinecek_kitap = kitap
                break

            if silinecek_kitap:
                self.envanter.remove(silinecek_kitap)
                self.toplam_kitap -= 1
                print(f"'{silinecek_kitap.get_ad()}' silindi.")

            else:
                print(f"'{kitap_adi}' adında bir kitap bulunamadı.")

    def listele(self):
        print("\n---- Kütüphane Envanteri ----")
        if not self.envanter:
            print("Kütüphanede kitap yok.")
            return

        print(f"{'No':<4}{'Kitap Adı':<25}{'Yazar':<20}{'Durum':<15}")
        print("-" * 64)
        for i, kitap in enumerate(self.envanter):
            print(f"{i+1:<4}{str(kitap)}")

    def odunc_kitap_ver(self, kitap_adi, kullanici_nesnesi):
        if not isinstance(kullanici_nesnesi, Person):
            print("HATA: Geçerli bir kullanıcı nesnesi gerekli.")
            return

        for kitap in self.envanter:

            if kitap.get_ad().lower() == kitap_adi.lower():
                if kitap._durum == "Müsait":
                     kitap.set_durum("Ödünçte")
                     kullanici_nesnesi.odunc_al(kitap_adi)
                     print(f"'{kitap.get_ad()}' {kullanici_nesnesi._ad} kişisine ödünç verildi.")
                     return
                else:
                     print(f"HATA: '{kitap.get_ad()}' zaten ödünçte.")
                     return
        print(f"HATA: '{kitap_adi}' adında bir kitap bulunamadı.")

    def odunc_kitap_geri_al(self, kitap_adi):
        for kitap in self.envanter:
            if kitap.get_ad().lower() == kitap_adi.lower():
                if kitap._durum == "Ödünçte":
                    kitap.set_durum("Müsait")
                    print(f"'{kitap.get_ad()}' iade edildi.")
                    return
                else:
                    print(f"HATA: '{kitap.get_ad()}' zaten kütüphanede.")
                    return
        print(f"HATA: '{kitap_adi}' adında bir kitap bulunamadı.")
    


            

    
```

```python
def ana_menu(library_manager, test_user):

    print(f"\nKullanıcı Girişi: {test_user.display_info()}")
    
    while True:
        print("\n" + "-"*80)
        print("---- KÜTÜPHANE YÖNETİMİ ----")
        print("-"*80)
        print("1- Kitap ekle")
        print("2- Kitap sil")
        print("3- Listele")
        print("4- Ödünc Kitap Ver ")
        print("5- Ödünc Kitap Geri Al")
        print(f"6- '{test_user._ad}' Ödünç Aldığı Kitapları Gör")
        print("7- Çıkış")
        
        secim = input("Lütfen yapmak istediğiniz işlemi seçin (1-7): ")
        
        if secim == '1':
            ad = input("Kitap Adı: "); yazar = input("Yazar: ")
            try:
                sayfa = int(input("Sayfa Sayısı: ")); yil = int(input("Yayın Yılı: "))
                yeni_kitap = Book(ad, yazar, sayfa, yil)
                library_manager.kitap_ekle(yeni_kitap)
            except ValueError:
                print("Sayfa ve Yıl sayı olmalıdır.")
        elif secim == '2':
            ad = input("Silinecek Kitabın Adı: ")
            library_manager.kitap_sil(ad)
        elif secim == '3':
            library_manager.listele()
        elif secim == '4':
            ad = input("Ödünç Verilecek Kitabın Adı: ")
            library_manager.odunc_kitap_ver(ad, test_user)
        elif secim == '5':
            ad = input("Geri Alınacak Kitabın Adı: ")
            library_manager.odunc_kitap_geri_al(ad, test_user)
        elif secim == '6':
             print(f"'{test_user._ad}' Aldığı Kitaplar: {test_user._kitaplar}")
        elif secim == '7':
            print("Sistemden çıkış yapılıyor. Güle güle!")
            break
        else:
            print("Geçersiz seçim.")
```
